```python
#FILTRANDO DADOS NULOS UTILIZANDO A MÉDIA

#biblioteca que permite dados nulos
import numpy as np

#criando um dataframe (df_salarios) para teste com as colunas nome e salário
#np.nan->são os dados nulos
df_salarios = pd.DataFrame({
    "nome":["Lucas","Juliana","Raphael","Junior","Satie"],
    "salario":[np.nan,4500,5000,np.nan,100000]
  })

'''
criando a coluna (salario_media) e recebendo a fórmula .fillna para preencher os campos nulos
df_salarios -> base de dados
salario_media -> coluna nova
dentro da base na coluna média iremos prencher os valores nulos(fill.na) com a média ('salario'.mean())
.round2 serve para fazer o arrendodamento do valor
criando a coluna salario_mediana para prencher os campos nulos com o median
.median serve também para calcular a média porém se tiver um valor fora do padrão ele desconsidera da média
'''

df_salarios["salario_media"] = df_salarios['salario'].fillna(df_salarios['salario'].mean().round(2))

df_salarios["salario_mediana"] = df_salarios['salario'].fillna(df_salarios['salario'].median())
df_salarios
```

```python
#ALTERANDO VALORES NULOS PELO VALOR SUPERIOR E ANTERIOR

#criando o dataframe temperatura com as colunas dia e temperatura
df_temperaturas = pd.DataFrame({
    "Dia": ['Segunda','Terça','Quarta','Quinta','Sexta','Sábado','Domingo'],
    "Temperatura": [30, np.nan, np.nan, 28, 27, np.nan, 26]
})

#criando a coluna preenchido_ffil com a funcão de caso tenha um valor nulo preenche com o valor anterior .ffill()
#criando a coluna preenchido_bfil com a função de caso tenha um valor nulo preenche com o valor o próximo valor .bfill()
df_temperaturas["preenchido_ffill"] = df_temperaturas["Temperatura"].ffill()
df_temperaturas["preenchido_bfill"] = df_temperaturas["Temperatura"].bfill()

print(df_temperaturas)
```

```python
#ALTERANDO VALORES NULOS PARA STRING 'Não informado'

#criando o dataframe df_cidades com as colunas nome e cidade
df_cidades = pd.DataFrame({
    'nome':['Lucas','Juliana','Raphael','Junior','Satie'],
    'cidade':['SP',np.nan,'BH','RJ',np.nan]
})
#adicionando ao dataframe uma coluna cidade_preenchida
#os valores nulos ficaram como "Não Informado"
df_cidades['cidade_preenchida'] = df_cidades['cidade'].fillna('Não informado')
df_cidades
```

```python
#criando uma váriavel df_limpo que recebe nosso df e utiliza a função dropna() para excluir campos nulos
df_limpo = df.dropna()
#somar e exibir todos os valores da variavel df_limpo que são nulos no caso 0 por conta da função anterior
df_limpo.isnull().sum()
```

```python
#exibindo o dataframe tratado, repare que o ano está no valor float
df_limpo.head()
```

```python
#aqui conseguimos realmente verificar que o tipo do dado ano está em float
df_limpo.info()
```

```python
#ALTERANDO TIPO DO DADO
#nossa base de dados df_limpo recebe df_limpo novamente mas será reconfigurado(assign) alterando a coluna ano
#no caso ele pega o ano antigo e modifica o tipo(.astype) pra int64
df_limpo = df_limpo.assign(ano = df_limpo['ano'].astype('int64'))
df_limpo
```

# Aula **3** - Visualização de códigos

```python
#chamando os valores da coluna senioridade do df para usar o comando plot
#o comando plot é responsável por gerar um grafico exigindo 2 necessidades
#kind = 'tipo da grafico' #title = 'titulo do grafico'
df_limpo['senioridade'].value_counts().plot(kind = 'bar',title ='Distribuição de senioridade')
```

```python
#importando biblioteca responsavel por
import seaborn as sns
```

```python
#utilizando o comando bar.plot para exibir um gráfico de barras
sns.barplot(data=df_limpo, x = 'senioridade', y ='usd')
```

```python
import matplotlib.pyplot as plt
```

```python
#definindo o tamanho da figura
plt.figure(figsize=(8,5))
#gerando o gráfico igual o anterior e incluindo informações para melhor entendimento dos dados
sns.barplot(data=df_limpo, x = 'senioridade', y ='usd')
#adicionando um titulo na imagem
plt.title('Salário médio por nivél de senioridade')
#adicionando um label pra nosso eixo X
plt.xlabel('Senioridade')
#adicionando um label para nosso eixo Y
plt.ylabel('Salário médio anual (USD)')
#exiba a imagem
plt.show()
```

```python
#deixando os dados do maior para o menor para melhor visualização
#utilizando o método groupby para agrupamento e ordenação (do maior pro menor)
#1 parametro do groupby coluna que vamos visualizar ('senioridade')
#2 parametro coluna que calcularemos as estatiticas ['usd']
#3 parametro .mean() para calcular a média
#4 parametro método sort_values para fazer ordenação
#5 parametro que o sort_values exige para ordenar do menor pro maior exigindo o parametro ascending = False
df_limpo.groupby('senioridade')['usd'].mean().sort_values(ascending=False)
```

```python
#adicionando o comando anterior na variavel ordem
#o .index no final é necessário para armazenar o nome de cada linha do dataframe ordenado
ordem = df_limpo.groupby('senioridade')['usd'].mean().sort_values(ascending=False).index
ordem
```

```python
#adicionando um novo parametro ao bar plot, order = ordem(váriavel que criamos anteriormente)
#desse jeito o gráfico exibe os dados do maior pro menor
plt.figure(figsize=(8,5))
sns.barplot(data=df_limpo, x = 'senioridade', y ='usd',order=ordem)
plt.title('Salário médio por nivél de senioridade')
plt.xlabel('Senioridade')
plt.ylabel('Salário médio anual (USD)')
plt.show()
```

```python
#verificando a destribuição salarial
#utilizando o estograma como plot ao invés do gráfico de barra
plt.figure(figsize=(10,5))
#criando o estograma com a função histplot
#1 parametro coluna que queremos visualizar
#2 parametro bins define o intervalo entre as barras que serão geradas no estograma
#3 parametro kde é uma linha adicionada ao gráfico com infos para melhor compreensão
sns.histplot(df_limpo['usd'],bins = 50, kde=True)
#definindo o titulo do estograma
plt.title('Distribuição dos salários anuais')
#definindo a label do eixo X
plt.xlabel('Salário em USD')
#definindo a label do eixo Y
plt.ylabel('Frequência')
#exibindo o gráfico
plt.show()
```

```python
#utilizando o comando boxplot para visualizar os valores da destribuição de forma diferente
#a linha do meio na caixa é a mediana (entre 100.000 e 200.000)
#os pontinhos ao lado direito são outliers (dados que são descrepantes)
plt.figure(figsize=(8,4))
sns.boxplot(x=df_limpo['usd'])
plt.title('Boxplot salário')
plt.xlabel('Salário em USD')
plt.show()
```

```python
#criando a variável ordem_senioridade para receber uma lista com os niveis
ordem_senioridade = ['Junior','Medio','Senior','Executivo']
#definindo o tamanho da figura
plt.figure(figsize=(10,5))
#chamando o sns e utilizando o boxplot novamente
sns.boxplot(x='senioridade',y='usd',data=df_limpo,order=ordem_senioridade)
plt.title('Boxplot da distribuição por senioridade')
plt.xlabel('Salário em USD')
plt.show()
```
